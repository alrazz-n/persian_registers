```python

import json
import gzip
import zstandard as zstd

from pathlib import Path
from collections import Counter

# ============================================================
# CONFIGURATION
# ============================================================

FILE = Path(r"C:\Users\alrazz\Downloads\10_1.jsonl(2)\10_1.jsonl(2)")

TEXT_COLUMN = "text"
OUTPUT_COLUMN = "new_web_register"

# Set this to False if auditing an original file
# that does not contain predictions.
CHECK_PREDICTIONS = False

# Set this to True if you know the expected labels.
CHECK_LABELS = False

EXPECTED_LABELS = [
    # "label_1",
    # "label_2",
]

# ============================================================
# OPEN FILE AS A STREAM
# ============================================================

def open_jsonl(path):

    raw = open(path, "rb")

    if path.name.endswith(".zst"):

        reader = zstd.ZstdDecompressor().stream_reader(
            raw,
            read_across_frames=True
        )

        import io

        return io.TextIOWrapper(
            reader,
            encoding="utf-8"
        )

    if path.name.endswith(".gz"):

        raw.close()

        return gzip.open(
            path,
            "rt",
            encoding="utf-8"
        )

    return open(
        path,
        "r",
        encoding="utf-8"
    )


# ============================================================
# AUDIT
# ============================================================

def audit_file(path):

    print("=" * 70)
    print("FILE INTEGRITY AUDIT")
    print("=" * 70)

    print("File:", path)
    print("Size (GB):", round(path.stat().st_size / 1024**3, 3))

    total_records = 0
    invalid_json = 0
    invalid_objects = 0
    missing_text = 0
    empty_text = 0
    missing_predictions = 0
    invalid_predictions = 0

    column_counts = Counter()
    first_record = None
    last_record = None

    reference_columns = None
    inconsistent_columns = 0

    try:

        with open_jsonl(path) as f:

            for line_number, line in enumerate(f, 1):

                if not line.strip():
                    continue

                # ------------------------------------------------
                # JSON VALIDATION
                # ------------------------------------------------

                try:

                    record = json.loads(line)

                except Exception as e:

                    invalid_json += 1

                    print(
                        f"\nINVALID JSON at line {line_number}: {e}"
                    )

                    continue

                if not isinstance(record, dict):

                    invalid_objects += 1
                    continue

                total_records += 1

                if first_record is None:
                    first_record = record

                last_record = record

                # ------------------------------------------------
                # COLUMN VALIDATION
                # ------------------------------------------------

                columns = set(record.keys())

                column_counts.update(columns)

                if reference_columns is None:

                    reference_columns = columns

                elif columns != reference_columns:

                    inconsistent_columns += 1

                # ------------------------------------------------
                # TEXT VALIDATION
                # ------------------------------------------------

                if TEXT_COLUMN not in record:

                    missing_text += 1

                elif not isinstance(record[TEXT_COLUMN], str):

                    missing_text += 1

                elif not record[TEXT_COLUMN].strip():

                    empty_text += 1

                # ------------------------------------------------
                # PREDICTION VALIDATION
                # ------------------------------------------------

                if CHECK_PREDICTIONS:

                    if OUTPUT_COLUMN not in record:

                        missing_predictions += 1

                    else:

                        probabilities = record[OUTPUT_COLUMN]

                        if not isinstance(probabilities, dict):

                            invalid_predictions += 1

                        else:

                            for label, probability in probabilities.items():

                                if (
                                    isinstance(probability, bool)
                                    or not isinstance(
                                        probability,
                                        (int, float)
                                    )
                                    or not 0 <= probability <= 1
                                ):

                                    invalid_predictions += 1
                                    break

                            if (
                                CHECK_LABELS
                                and set(probabilities.keys())
                                != set(EXPECTED_LABELS)
                            ):

                                invalid_predictions += 1

                # ------------------------------------------------
                # PROGRESS
                # ------------------------------------------------

                if total_records % 1_000_000 == 0:

                    print(
                        f"Checked {total_records:,} records..."
                    )

    except Exception as e:

        print("\nFATAL FILE ERROR:", repr(e))

        raise

    # ============================================================
    # REPORT
    # ============================================================

    print("\n" + "=" * 70)
    print("AUDIT RESULTS")
    print("=" * 70)

    print("Total valid JSON records:", f"{total_records:,}")
    print("Invalid JSON lines:", invalid_json)
    print("Non-object JSON records:", invalid_objects)

    print("Missing text:", missing_text)
    print("Empty text:", empty_text)

    if CHECK_PREDICTIONS:

        print("Missing predictions:", missing_predictions)
        print("Invalid predictions:", invalid_predictions)

    print("Records with inconsistent columns:", inconsistent_columns)

    print("\nCOLUMNS FOUND:")

    for column, count in column_counts.most_common():

        print(
            f"{column}: {count:,} records"
        )

    if first_record:

        print("\nFIRST RECORD:")

        print(
            json.dumps(
                first_record,
                ensure_ascii=False,
                indent=2
            )[:2000]
        )

    if last_record:

        print("\nLAST RECORD:")

        print(
            json.dumps(
                last_record,
                ensure_ascii=False,
                indent=2
            )[:2000]
        )

    # ============================================================
    # FINAL STATUS
    # ============================================================

    problems = (
        invalid_json
        + invalid_objects
        + missing_text
        + missing_predictions
        + invalid_predictions
        + inconsistent_columns
    )

    print("\n" + "=" * 70)

    if problems == 0:

        print("AUDIT PASSED!")

        print("No problems detected in the checked fields.")

    else:

        print("AUDIT FAILED!")

        print("Problems detected:", problems)

    print("=" * 70)

    return problems == 0


# ============================================================
# RUN
# ============================================================

if not FILE.exists():

    raise FileNotFoundError(FILE)

audit_file(FILE)
```

```python

import json
from pathlib import Path
from collections import Counter, defaultdict

FILE = Path(
    r"C:\Users\alrazz\Downloads\10_1.jsonl(2)\10_1.jsonl(2)"
)

column_counts = Counter()
schema_counts = Counter()
missing_examples = defaultdict(list)

total = 0
reference_columns = None

with open(FILE, "r", encoding="utf-8") as f:

    for line_number, line in enumerate(f, 1):

        if not line.strip():
            continue

        record = json.loads(line)

        if not isinstance(record, dict):
            continue

        total += 1

        columns = frozenset(record.keys())

        schema_counts[columns] += 1
        column_counts.update(columns)

        if reference_columns is None:
            reference_columns = set(columns)

        missing = reference_columns - set(columns)
        extra = set(columns) - reference_columns

        if missing or extra:

            for column in missing:
                if len(missing_examples[column]) < 3:
                    missing_examples[column].append(line_number)

print("=" * 70)
print("SCHEMA DIAGNOSTICS")
print("=" * 70)

print("Total records:", total)

print("\nCOLUMN FREQUENCIES:")

for column, count in column_counts.most_common():

    print(
        f"{column}: {count:,} "
        f"({count / total * 100:.2f}%)"
    )

print("\nMISSING COLUMNS:")

for column in sorted(reference_columns):

    missing_count = total - column_counts[column]

    if missing_count:

        print(
            f"{column}: missing in {missing_count:,} records"
        )

        print(
            "Example line numbers:",
            missing_examples[column]
        )

print("\nDISTINCT SCHEMAS:")

for schema, count in schema_counts.most_common():

    print(f"\nRecords: {count:,}")

    print("Columns:", sorted(schema))

print("\nTOTAL DISTINCT SCHEMAS:", len(schema_counts))
```
