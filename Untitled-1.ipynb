```python
# ============================================
# ARIABERT TOKENIZATION CHECK
# ============================================

from datasets import load_from_disk
from transformers import AutoTokenizer
import numpy as np

DATASET_PATH = r"C:\Users\alrazz\Documents\GitHub\persian_registers\Register\Dataset_Hugging_face\without_NA\Combined_hybrid_no_NA"

# Load dataset
dataset = load_from_disk(DATASET_PATH)

# Load AriaBERT tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "ViraIntelligentDataMining/AriaBERT"
)

print("=== TOKENIZER CHECK ===")
print("Tokenizer class:", type(tokenizer))
print("Vocab size:", tokenizer.vocab_size)
print("Model max length:", tokenizer.model_max_length)

# Check all splits
for split in ["train", "dev", "test"]:

    print(f"\n=== {split.upper()} ===")

    texts = dataset[split]["text"]

    token_lengths = []

    for text in texts:
        encoded = tokenizer(
            text,
            truncation=False,
            padding=False,
        )

        token_lengths.append(len(encoded["input_ids"]))

    token_lengths = np.asarray(token_lengths)

    print("Number of examples:", len(token_lengths))
    print("Min tokens:", token_lengths.min())
    print("Mean tokens:", token_lengths.mean())
    print("Median tokens:", np.median(token_lengths))
    print("Max tokens:", token_lengths.max())

    print(
        "Over 512:",
        (token_lengths > 512).sum(),
        f"({(token_lengths > 512).mean()*100:.2f}%)"
    )
```

```python
aria_tokenizer = AutoTokenizer.from_pretrained(
    "ViraIntelligentDataMining/AriaBERT"
)

xlm_tokenizer = AutoTokenizer.from_pretrained(
    "xlm-roberta-large"
)
```

```python
import numpy as np

texts = dataset["train"]["text"]

for name, tok in [
    ("AriaBERT", aria_tokenizer),
    ("XLM-R", xlm_tokenizer),
]:
    lengths = []

    for text in texts:
        ids = tok(
            text,
            truncation=False,
            add_special_tokens=True
        )["input_ids"]

        lengths.append(len(ids))

    lengths = np.asarray(lengths)

    print(f"\n{'='*50}")
    print(name)
    print("="*50)
    print("Mean tokens:", lengths.mean())
    print("Median tokens:", np.median(lengths))
    print("Max tokens:", lengths.max())
    print("Over 512:", (lengths > 512).sum())
    print("Over 512 %:", (lengths > 512).mean() * 100)
```

```python
text = dataset["train"][0]["text"]

for name, tok in [
    ("AriaBERT", aria_tokenizer),
    ("XLM-R", xlm_tokenizer),
]:
    encoded = tok(
        text,
        truncation=False,
        add_special_tokens=True
    )

    tokens = tok.convert_ids_to_tokens(encoded["input_ids"])

    print("\n", name)
    print("Number of tokens:", len(tokens))
    print("First 50 tokens:")
    print(tokens[:50])
```

```python
text = dataset["train"][0]["text"]

for name, tok in [
    ("AriaBERT", aria_tokenizer),
    ("XLM-R", xlm_tokenizer),
]:
    encoded = tok(
        text,
        truncation=True,
        max_length=512,
        add_special_tokens=True,
    )

    decoded = tok.decode(
        encoded["input_ids"],
        skip_special_tokens=True
    )

    print("\n" + "="*60)
    print(name)
    print("="*60)

    print("Original characters:", len(text))
    print("512-token decoded characters:", len(decoded))
    print("Coverage:", len(decoded) / len(text) * 100, "%")
    print("Last 200 decoded characters:")
    print(decoded[-200:])
```

```python
from transformers import AutoModelForSequenceClassification

models = {
    "BGE-M3-RetroMAE": "BAAI/bge-m3",
    "XLM-RoBERTa-large": "FacebookAI/xlm-roberta-large",
}

for name, path in models.items():
    model = AutoModelForSequenceClassification.from_pretrained(
        path,
        num_labels=2
    )

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)

    print(f"\n{name}")
    print(f"Total parameters:     {total:,}")
    print(f"Total parameters (M): {total / 1e6:.2f}")
    print(f"Trainable parameters: {trainable:,}")
```

# Checking training data disribution

```python
import json
import pandas as pd
import numpy as np

# Path to your JSONL file
file_path = r"C:\Users\alrazz\Documents\GitHub\persian_registers\Register\jsonl_output\without_NA\Combined_single_no_NA.jsonl"

# Read the JSONL file
df = pd.read_json(file_path, lines=True)

# Count words in each text using whitespace
df["word_count"] = df["text"].fillna("").apply(
    lambda text: len(text.split())
)

# Calculate descriptive statistics
word_stats = df["word_count"].describe()

print("=== Word Count Statistics ===")
print(word_stats)

print("\n=== Additional Statistics ===")
print(f"Total documents: {len(df)}")
print(f"Total words: {df['word_count'].sum()}")
print(f"Mean words per document: {df['word_count'].mean():.2f}")
print(f"Median words per document: {df['word_count'].median():.2f}")
print(f"Minimum words: {df['word_count'].min()}")
print(f"Maximum words: {df['word_count'].max()}")
print(f"Standard deviation: {df['word_count'].std():.2f}")
print(f"Variance: {df['word_count'].var():.2f}")

# Additional percentiles
print("\n=== Percentiles ===")
for p in [10, 25, 50, 75, 90, 95, 99]:
    print(f"{p}th percentile: {df['word_count'].quantile(p / 100):.2f}")

# Inspect documents with the shortest and longest texts
print("\n=== Shortest Documents ===")
print(df.nsmallest(5, "word_count")[["id", "word_count"]])

print("\n=== Longest Documents ===")
print(df.nlargest(5, "word_count")[["id", "word_count"]])

# Save document-level word counts
#df.to_csv("document_word_counts.csv", index=False)
```
