```python
import os
import pickle

import numpy as np
import pandas as pd

from scipy.stats import pearsonr, spearmanr

import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
INPUT = "score_distributions.pkl"

OUTDIR = "."
os.makedirs(OUTDIR, exist_ok=True)


titles = [
    "WDS score",
    "lang_score",
    "url_score",
    "punctuation_score",
    "singular_chars_score",
    "numbers_score",
    "repeated_score",
    "informativeness_score",
    "n_long_segment_score",
    "great_segment_score",
]
```

```python
with open(INPUT, "rb") as f:
    data = pickle.load(f)
```

```python
results = []

fig = make_subplots(
    rows=5,
    cols=2,
    subplot_titles=titles,
)

for i, title in enumerate(titles):

    row = i // 2 + 1
    col = i % 2 + 1

    bins = sorted(data["all"][i].keys())

    score_bins = []
    prevalence = []

    for b in bins:

        all_count = data["all"][i][b]
        junk_count = data["junk"][i].get(b, 0)

        if all_count == 0:
            continue

        score_bins.append(b)
        prevalence.append(junk_count / all_count)

    score_bins = np.asarray(score_bins, dtype=float)
    prevalence = np.asarray(prevalence)
```

```python
    pearson_r, pearson_p = pearsonr(
        score_bins,
        prevalence,
    )

    spearman_rho, spearman_p = spearmanr(
        score_bins,
        prevalence,
    )

    results.append(
        {
            "score": title,
            "pearson_r": pearson_r,
            "pearson_p": pearson_p,
            "spearman_rho": spearman_rho,
            "spearman_p": spearman_p,
        }
    )
```
