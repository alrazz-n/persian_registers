```python
import json
import io
import numpy as np
import zstandard as zstd

junk_file = r"C:\Users\alrazz\annotated_data\junk_records\10_1.junk.jsonl.zst"

scores = []

dctx = zstd.ZstdDecompressor()

with open(junk_file, "rb") as fh:
    with dctx.stream_reader(fh) as reader:
        text_stream = io.TextIOWrapper(reader, encoding="utf-8")

        for line in text_stream:
            doc = json.loads(line)
            scores.append(doc["doc_scores"])

scores = np.array(scores)

print(scores.shape)
print("Mean:")
print(scores.mean(axis=0))

print("\nMedian:")
print(np.median(scores, axis=0))

print("\nStd:")
print(scores.std(axis=0))
```
