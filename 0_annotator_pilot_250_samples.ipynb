```python
labels_structure = {
    "MT": [],
    "LY": [],
    "SP": ["it", "os"],
    "ID": [],
    "NA": ["ne", "sr", "nb", "on"],
    "HI": ["re", "oh"],
    "IN": ["en", "ra", "dtp", "fi", "lt", "oi"],
    "OP": ["rv", "ob", "rs", "av", "oo"],
    "IP": ["ds", "ed", "oe"],
}
```

```python
import random
import numpy as np

random.seed(42)

target_counts = {
    label: TARGET_PER_LABEL
    for label in leaf_labels
}

current_counts = {
    label: 0
    for label in leaf_labels
}

selected_indices = set()

available_indices = set(df.index)


def useful_labels(idx):
    """Labels in this document that still need coverage."""
    
    labels = document_labels[idx]
    
    return {
        label
        for label in labels
        if label in target_counts
        and current_counts[label] < target_counts[label]
    }


while True:

    # Find documents that can still improve coverage
    candidates = []

    for idx in available_indices - selected_indices:

        useful = useful_labels(idx)

        if useful:
            candidates.append((idx, useful))

    if not candidates:
        break

    # Find maximum number of useful labels
    max_score = max(
        len(useful)
        for idx, useful in candidates
    )

    best_candidates = [
        (idx, useful)
        for idx, useful in candidates
        if len(useful) == max_score
    ]

    # Random tie-breaking
    idx, useful = random.choice(best_candidates)

    selected_indices.add(idx)

    # Update coverage counts
    for label in useful:
        current_counts[label] += 1

    # Stop once every label reaches target
    if all(
        current_counts[label] >= target_counts[label]
        for label in target_counts
    ):
        break


print("Documents selected for coverage:", len(selected_indices))
print(current_counts)
```

```python
from collections import Counter
import pandas as pd
import random

random.seed(42)

N_SAMPLE = 250
TARGET = 15

parent_only = ["MT", "LY", "ID"]

leaf_labels = [
    child
    for children in labels_structure.values()
    for child in children
]

target_labels = parent_only + leaf_labels

print("Number of target labels:", len(target_labels))
```

```python
def parse_labels(x):
    if pd.isna(x):
        return set()

    return {
        label.strip()
        for label in str(x).split(";")
        if label.strip() and label.strip() != "-"
    }


df_sampling = df_sampling.copy()

df_sampling["all_labels"] = (
    df_sampling["Turku_NLP"]
    .apply(parse_labels)
)

df_sampling["leaf_set"] = (
    df_sampling["Turku_NLP_sub"]
    .apply(parse_labels)
)
```

```python
def get_sampling_labels(row):
    labels = set()

    # Parent-only labels
    labels.update(
        row["all_labels"] & set(parent_only)
    )

    # Leaf labels
    labels.update(
        row["leaf_set"] & set(leaf_labels)
    )

    return labels


df_sampling["sampling_labels"] = (
    df_sampling.apply(
        get_sampling_labels,
        axis=1
    )
)
```

```python
observed_sampling_labels = set()

for labels in df_sampling["sampling_labels"]:
    observed_sampling_labels.update(labels)

print(sorted(observed_sampling_labels))

print(len(sorted(observed_sampling_labels)))
```

```python
minimum_coverage = {
    label: TARGET
    for label in target_labels
}

current_counts = {
    label: 0
    for label in target_labels
}
```

```python
available_indices = set(df_sampling.index)
selected_indices = set()


def document_score(labels):
    score = 0.0

    for label in labels:

        if label not in current_counts:
            continue

        remaining = (
            TARGET - current_counts[label]
        )

        if remaining > 0:
            score += remaining / TARGET

    return score
```

```python
while True:

    candidates = []

    for idx in available_indices - selected_indices:

        labels = df_sampling.loc[
            idx, "sampling_labels"
        ]

        score = document_score(labels)

        if score > 0:
            candidates.append(
                (idx, score)
            )

    # Nothing else can improve coverage
    if not candidates:
        break

    max_score = max(
        score for _, score in candidates
    )

    best_candidates = [
        (idx, score)
        for idx, score in candidates
        if score == max_score
    ]

    idx, _ = random.choice(best_candidates)

    selected_indices.add(idx)

    labels = df_sampling.loc[
        idx, "sampling_labels"
    ]

    for label in labels:

        if label in current_counts:

            if current_counts[label] < TARGET:
                current_counts[label] += 1

    # Stop when every target is covered
    if all(
        current_counts[label] >= TARGET
        for label in target_labels
    ):
        break
```

```python
coverage = pd.DataFrame({
    "target": minimum_coverage,
    "selected": current_counts
})

coverage["target_met"] = (
    coverage["selected"] >= coverage["target"]
)

print(coverage.sort_index())
```
